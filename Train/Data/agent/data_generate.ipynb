```python
import json
import random
from tqdm import tqdm
from data_utils import (
    drink_order_data,
    weather_data,
    hot_search_data,
    horoscope_data
)

# 输出文件路径
OUTPUT_PATH = "Train/Data/agent/agent_final_dataset.json"
data_count = 60
final_dataset = []
```

```python
DRINK_INPUT_PATH = "Train/Data/agent/drink_query_pairs.jsonl"

def load_drink_query_pairs(path):
    data = []
    with open(path, "r", encoding="utf-8") as f:
        for line in f:
            obj = json.loads(line.strip())
            data.append(obj)
    return data

def build_drink_data(pairs, n=60):
    results = []
    sampled = random.sample(pairs, n)

    for item in tqdm(sampled, desc="Generating drink data"):
        query = item["query"]
        attributes = item["attributes"]

        output = drink_order_data(query, attributes)

        results.append({
            "type": "drink",
            "input": {
                "query": query,
                "attributes": attributes
            },
            "output": output
        })
    return results

drink_pairs = load_drink_query_pairs(DRINK_INPUT_PATH)
final_dataset += build_drink_data(drink_pairs, data_count)
print(final_dataset[-1])
```

```python
def build_weather_data(n=60):
    results = []
    for _ in tqdm(range(n), desc="Generating weather data"):
        output = weather_data()
        print(output)
        results.append({
            "type": "weather",
            "input": None,     # queries inside weather_data() are auto-generated
            "output": output
        })
    return results
final_dataset += build_weather_data(data_count)
print(final_dataset[-1])
```

```python
def build_hot_search_data(n=60):
    results = []
    for _ in tqdm(range(n), desc="Generating hot-search data"):
        output = hot_search_data()
        results.append({
            "type": "hot_search",
            "input": None,
            "output": output
        })
    return results
final_dataset += build_hot_search_data(data_count)
print(final_dataset[-1])
```

```python
def build_horoscope_data(n=60):
    results = []
    for _ in tqdm(range(n), desc="Generating horoscope data"):
        output = horoscope_data()
        results.append({
            "type": "horoscope",
            "input": None,
            "output": output
        })
    return results
final_dataset += build_horoscope_data(data_count)
print(final_dataset[-1])
```

```python
with open(OUTPUT_PATH, "w", encoding="utf-8") as f:
    for item in final_dataset:
        f.write(json.dumps(item, ensure_ascii=False) + "\n")

print(f"✅ 数据集已生成，共 {len(final_dataset)} 条，已写入：{OUTPUT_PATH}")
```
